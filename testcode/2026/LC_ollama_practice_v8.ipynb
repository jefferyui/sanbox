````python
from langchain_core.prompts import ChatPromptTemplate
from langchain_ollama import ChatOllama
from langchain_core.output_parsers import StrOutputParser

prompt = ChatPromptTemplate.from_messages([
    ("system","你是一位資料科學家"),
    ("human","{question}")
])

llm = ChatOllama(
    model="qwen3.5:4b",
    temperature=0.2
)

chain = prompt | llm | StrOutputParser()

while True:

    q = input("你：")

    if q == "exit":
        break

    ans = chain.invoke(
        {
            "question": q
        }
    )

    print("AI：", ans)
````

````python
from langchain_community.document_loaders import WebBaseLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_ollama import OllamaEmbeddings, ChatOllama
from langchain_chroma import Chroma

##################################################
# Step1 載入文章
##################################################

loader = WebBaseLoader(
    "https://lilianweng.github.io/posts/2023-06-23-agent/"
)

docs = loader.load()

print(f"文章數量：{len(docs)}")

##################################################
# Step2 切 Chunk
##################################################

splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=100
)

documents = splitter.split_documents(docs)

print(f"Chunk數量：{len(documents)}")

##################################################
# Step3 Embedding
##################################################

embedding = OllamaEmbeddings(
    model="nomic-embed-text"
)

##################################################
# Step4 建立向量資料庫
##################################################

db = Chroma.from_documents(
    documents=documents,
    embedding=embedding
)

##################################################
# Step5 建立 Retriever
##################################################

retriever = db.as_retriever(
    search_kwargs={"k":3}
)

##################################################
# Step6 建立 LLM
##################################################

llm = ChatOllama(
    model="qwen3.5:4b",
    temperature=0
)

##################################################
# Step7 詢問
##################################################

question = "What is an AI Agent?"

docs = retriever.invoke(question)

context = "\n\n".join(
    doc.page_content
    for doc in docs
)

prompt = f"""
請依照下面資料回答。

資料：

{context}

問題：

{question}
"""

answer = llm.invoke(prompt)

print("="*60)
print(answer.content)
````

````python
import logging
from langchain_core.documents import Document
from langchain_chroma import Chroma
from langchain_ollama import OllamaEmbeddings
from langchain_classic.retrievers import MultiQueryRetriever
from langchain_ollama import ChatOllama
from langchain_core.prompts import PromptTemplate

# 設定 Log 層級以觀察 MultiQueryRetriever 的運作
logging.basicConfig()
logging.getLogger(
    "langchain_classic.retrievers.multi_query"
).setLevel(logging.INFO)

# ==========================================
# 建立文件資料庫 (Documents)
# ==========================================
documents = [
    Document(
        page_content="""
CMP製程利用化學機械研磨平坦化晶圓表面。
常見問題包含刮傷、dishing、erosion。
"""
    ),
    Document(
        page_content="""
SPC利用管制圖監控製程變異。
常用方法包含Xbar-R chart、EWMA、CUSUM。
"""
    ),
    Document(
        page_content="""
Wafer defect包含particle、
scratch、pattern defect。

Particle污染可能造成電路短路、
開路或良率下降。
"""
    ),
    Document(
        page_content="""
降低晶圓缺陷需要進行設備與製程參數控制。

常見改善方法包含：
1. 監控設備腔體(Chamber)狀態，
避免溫度、壓力、RF power漂移。

2. 透過FDC(Fault Detection and Classification)
即時分析設備sensor資料，
提前發現異常狀況。

3. 定期設備保養(PM)，
降低particle產生與設備老化造成的缺陷。
"""
    ),
    Document(
        page_content="""
晶圓缺陷改善需要結合缺陷分析與製程最佳化。

常用方法包含：
1. Defect Pareto分析，
找出主要缺陷來源。

2. DOE實驗設計，
分析製程參數對缺陷率的影響。

3. Recipe最佳化，
調整溫度、壓力、時間與化學液比例，
降低製程變異並提升良率。
"""
    )
]

# ==========================================
# 初始化 Embedding 與向量資料庫 (Chroma)
# ==========================================
embedding = OllamaEmbeddings(
    model="nomic-embed-text"
)

db = Chroma.from_documents(
    documents,
    embedding
)

retriever = db.as_retriever(
    search_kwargs={
        "k": 2
    }
)

# ==========================================
# 初始化 LLM 與提示詞樣板 (QUERY_PROMPT)
# ==========================================
llm = ChatOllama(
    model="qwen3.5:4b",
    reasoning=False
)

QUERY_PROMPT = PromptTemplate(
    input_variables=["question"],
    template="""你是位半導體製程專家。

請將以下問題改寫成3個不同角度的搜尋問題。

規則：
1. 必須使用繁體中文
2. 不要使用簡體中文
3. 保留半導體專有名詞，例如：
   SPC、FDC、CMP、Wafer、Recipe
4. 每個問題一行輸出

原始問題：
{question}

輸出：
"""
)


# ==========================================
# 4. 建立 Retriever
# ==========================================
multi_retriever = MultiQueryRetriever.from_llm(
    retriever=retriever,
    llm=llm,
    prompt=QUERY_PROMPT
)

# ==========================================
# 5. 提問與檢索資料
# ==========================================
question = "如何降低晶圓缺陷"

# 找相關文件
results = multi_retriever.invoke(question)

# 去除重複文件
unique_docs = {}

for doc in results:
    unique_docs[
        doc.page_content
    ] = doc

results = list(unique_docs.values())

print("找到的資料:")
for r in results:
    print("----------------")
    print(r.page_content)

# ==========================================
# 6. 丟給 Ollama 生成最終回答
# ==========================================
context = "\n".join(
    r.page_content for r in results
)

prompt = f"""你是位半導體製程專家。

請仔細閱讀以下所有資料，
整合所有相關資訊回答問題。

要求：
1. 不要只引用第一筆資料
2. 將多筆資料融合合成完整答案
3. 如果資料不足，再說明不足

資料：
{context}

問題：
{question}

回答：
"""

llm = ChatOllama(
    model="qwen3.5:0.8b",
    temperature=0,
    reasoning=False
)

answer = llm.invoke(prompt)

print("\n==================")
print("AI回答:")
print(answer.content)
````

````python
from llama_index.core import (
    SimpleDirectoryReader,
    VectorStoreIndex,
    Settings
)

from llama_index.llms.ollama import Ollama

from llama_index.embeddings.ollama import OllamaEmbedding


from langchain_ollama import ChatOllama

from langchain_core.prompts import ChatPromptTemplate

from langchain_core.runnables import (
    RunnablePassthrough
)


# ==================================================
# 1. 設定 LlamaIndex
#    負責 RAG
# ==================================================


# Embedding Model

Settings.embed_model = OllamaEmbedding(
    model_name="nomic-embed-text"
)


# LLM

Settings.llm = Ollama(
    model="qwen3.5:0.8b",
    reasoning=False
)



# ==================================================
# 2. 讀取文件
# ==================================================

documents = SimpleDirectoryReader(
    "./RAG_files"
).load_data()



# ==================================================
# 3. 建立 LlamaIndex Index
# ==================================================

index = VectorStoreIndex.from_documents(
    documents
)



# 建立 Retriever

retriever = index.as_retriever(
    similarity_top_k=3
)



# ==================================================
# 4. LangChain LLM
# ==================================================

llm = ChatOllama(
    model="qwen3.5:0.8b",
    reasoning = False
)



# ==================================================
# 5. LangChain Prompt
# ==================================================

prompt = ChatPromptTemplate.from_template(
"""
你是一個半導體製程專家。

請根據以下文件回答問題。

文件內容:
{context}


問題:
{question}


如果文件沒有答案，
請回答不知道。

"""
)



# ==================================================
# 6. LlamaIndex Retriever
#    轉成 LangChain Runnable
# ==================================================

def retrieve_context(question):

    nodes = retriever.retrieve(
        question
    )

    context = "\n\n".join(
        [
            node.text
            for node in nodes
        ]
    )

    return context



# ==================================================
# 7. LangChain Chain
# ==================================================

chain = (

    {
        "context":
            lambda x:
            retrieve_context(
                x["question"]
            ),

        "question":
            RunnablePassthrough()

    }

    |

    prompt

    |

    llm

)



# ==================================================
# 8. 測試
# ==================================================
# response = chain.invoke(
#     {
#         "question":"CMP常見缺陷?"
#     }
# )
# print(response.content)

for chunk in chain.stream(
    {
        "question": "CMP常見缺陷?"
    }
):
    print(chunk.content, end="", flush=True)



# while True:


#     question = input(
#         "\n問題(exit離開): "
#     )


#     if question=="exit":
#         break



#     response = chain.invoke(
#         {
#             "question":question
#         }
#     )


#     print("\n================")
#     print(response.content)
#     print("================")
````

````python
import asyncio


from langchain_ollama import ChatOllama

from langchain.agents import create_agent


from langchain_mcp_adapters.client import (
    MultiServerMCPClient
)



async def main():


    ################################################
    # 連 MCP Server
    ################################################

    client = MultiServerMCPClient(
        {


            # "stock":{

            #     "transport":"stdio",

            #     "command":"python",

            #     "args":[
            #         "stock_server.py"
            #     ]

            # },
            "stock":{

                "transport":"streamable_http",

                "url":"http://127.0.0.1:8000/mcp"

                },

            # "weather":{

            #     "transport":"stdio",

            #     "command":"python",

            #     "args":[
            #         "weather_server.py"
            #     ]

            # }

        }
    )


    ################################################
    # 取得 MCP Tools
    ################################################

    tools = await client.get_tools()


    print("Tools:")

    for t in tools:

        print(
            t.name
        )



    ################################################
    # Ollama
    ################################################

    llm = ChatOllama(
        model="qwen3.5:0.8b",
        temperature=0,
        reasoning=False
    )


    agent = create_agent(
        model=llm,
        tools=tools
    )



    ################################################
    # 問問題
    ################################################

    response = await agent.ainvoke(
        {
            "messages":[

                {
                    "role":"user",
                    "content":
                    """
                    幫我查台積電2330.TW股價，
                    以及台北現在天氣
                    """
                }

            ]
        }
    )



    print("===================")

    print(
        response["messages"][-1].content
    )



# asyncio.run(main())
await main()
````

````python
# ============================================
# Milvus 已存在文件去重
# Similarity Threshold 去除高度相似文件
# ============================================

from pymilvus import MilvusClient
from sklearn.metrics.pairwise import cosine_similarity
from langchain_ollama import OllamaEmbeddings   
embedding = OllamaEmbeddings(model="nomic-embed-text")

# -----------------------------
# 1. 連線 Milvus Lite
# -----------------------------
client = MilvusClient(
    uri="./milvus_demo.db"
)
# client = MilvusClient(
#     uri="./milvus_demo.db"
# )

collection_name = "semiconductor_docs"
client.load_collection(
    collection_name="semiconductor_docs"
)

# -----------------------------
# 2. 取得 Milvus 所有文件
# -----------------------------
results = client.query(
    collection_name="semiconductor_docs",
    filter="",
    output_fields=[
        "pk",
        "text",
        "metadata"
    ],
    limit=10000
)

# results = client.query(
#     collection_name="semiconductor_docs",
#     filter="",
#     output_fields=[
#         "text",
#         "metadata"
#     ],
#     # limit=10
# )



print(f"目前文件數量: {len(results)}")


# -----------------------------
# 3. 取出文字
# -----------------------------
texts = [
    item["text"]
    for item in results
]


# -----------------------------
# 4. 重新計算 Embedding
# 注意：
# embedding 是你的 LangChain Embedding Model
# 例如 OllamaEmbeddings
# -----------------------------

vectors = embedding.embed_documents(texts)


# -----------------------------
# 5. 計算文件間相似度
# -----------------------------

similarity_matrix = cosine_similarity(
    vectors
)


# -----------------------------
# 6. 找重複文件
# threshold 越高越嚴格
#
# 0.98 = 幾乎完全相同
# 0.95 = 高度相似
# 0.90 = 可能相關
# -----------------------------

threshold = 0.95


duplicate_ids = []


for i in range(len(results)):

    for j in range(i + 1, len(results)):

        similarity = similarity_matrix[i][j]

        if similarity >= threshold:

            duplicate_id = results[j]["pk"]

            duplicate_ids.append(
                duplicate_id
            )

            print("=" * 50)
            print("發現重複文件")
            print(f"Similarity: {similarity:.4f}")

            print("\n保留:")
            print(results[i]["text"][:100])

            print("\n刪除:")
            print(results[j]["text"][:100])



print("\n需要刪除數量:", len(duplicate_ids))


# -----------------------------
# 7. 刪除重複文件
# -----------------------------

if duplicate_ids:

    client.delete(
        collection_name=collection_name,
        filter=f"pk in {duplicate_ids}"
    )

    print("刪除完成")

else:

    print("沒有發現重複文件")



# -----------------------------
# 8. 查看剩餘數量
# -----------------------------

stats = client.get_collection_stats(
    collection_name=collection_name
)

print("\n目前 DB 狀態:")
print(stats)
````

````python
import os
from langchain_community.utilities import SQLDatabase
from langchain_community.tools.sql_database.tool import QuerySQLDataBaseTool
# from langchain.chains import create_sql_query_chain
from langchain_classic.chains import create_sql_query_chain
from langchain_openai import ChatOpenAI
from operator import itemgetter
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import PromptTemplate
from langchain_core.runnables import RunnablePassthrough
from sqlalchemy import create_engine, Column, Integer, String, ForeignKey
from sqlalchemy.orm import declarative_base, sessionmaker
from langchain_ollama import ChatOllama

import re

def clean_sql(text):
    # 找 SQLQuery: 後面的內容
    match = re.search(
        r"SQLQuery:\s*(.*)",
        text,
        re.DOTALL | re.IGNORECASE
    )

    if match:
        text = match.group(1)

    # 移除 markdown code block
    text = text.replace("```sql", "")
    text = text.replace("```", "")

    return text.strip()

# ==========================================
# 步驟 1: 建立模擬資料庫 (SQLite)
# ==========================================
engine = create_engine("sqlite:///company.db", echo=False)
Base = declarative_base()

class Department(Base):
    __tablename__ = "departments"
    id = Column(Integer, primary_key=True)
    name = Column(String)

class Employee(Base):
    __tablename__ = "employees"
    id = Column(Integer, primary_key=True)
    name = Column(String)
    salary = Column(Integer)
    department_id = Column(Integer, ForeignKey("departments.id"))

Base.metadata.create_all(engine)

# 寫入測試資料
Session = sessionmaker(bind=engine)
session = Session()

if session.query(Department).count() == 0:
    hr = Department(name="人力資源部")
    tech = Department(name="技術研發部")
    sales = Department(name="業務部")
    session.add_all([hr, tech, sales])
    session.commit()

    session.add_all([
        Employee(name="張小明", salary=75000, department_id=tech.id),
        Employee(name="李華", salary=82000, department_id=tech.id),
        Employee(name="王五", salary=50000, department_id=hr.id),
        Employee(name="趙六", salary=60000, department_id=sales.id),
        Employee(name="陳七", salary=90000, department_id=sales.id),
    ])
    session.commit()
session.close()

# ==========================================
# 步驟 2: 初始化 LangChain SQLDatabase 物件
# ==========================================
db = SQLDatabase.from_uri("sqlite:///company.db")
# llm = ChatOpenAI(model="gpt-4o-mini", temperature=0)
llm = ChatOllama(
    model="qwen3.5:0.8b",
    temperature=0,
    reasoning=False
)
# ==========================================
# 步驟 3: 建立生成 SQL 與執行的 Chain
# ==========================================

# 1. 建立生成 SQL 的 Chain
# write_query_chain = create_sql_query_chain(llm, db)
write_query_chain = (
    create_sql_query_chain(llm, db)
    | clean_sql
)
# 2. 建立執行 SQL 的 Tool
execute_query = QuerySQLDataBaseTool(db=db)

# 3. 建立最後用自然語言整理回答的 Prompt
answer_prompt = PromptTemplate.from_template(
    """給定以下的使用者問題、對應的 SQL 查詢語法以及查詢結果，請用親切且專業的繁體中文回答使用者的問題。

使用者問題: {question}
SQL 查詢: {query}
SQL 執行結果: {result}

回答："""
)

# 4. 串接完整鏈路 (LCEL - LangChain Expression Language)
chain = (
    RunnablePassthrough.assign(query=write_query_chain)
    .assign(result=itemgetter("query") | execute_query)
    .assign(answer=answer_prompt | llm | StrOutputParser())
)

# ==========================================
# 步驟 4: 測試查詢
# ==========================================
if __name__ == "__main__":
    user_question = "請幫我找出技術研發部薪水最高的員工是誰？薪水是多少？"
    
    print(f"使用者問題: {user_question}\n")
    
    response = chain.invoke({"question": user_question})
    
    print(f"【生成的 SQL】\n{response['query']}\n")
    print(f"【資料庫回傳結果】\n{response['result']}\n")
    print(f"【最終自然語言回答】\n{response['answer']}")
````
