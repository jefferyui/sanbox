```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# 讀取灰階圖片
img = cv2.imread("j.png", cv2.IMREAD_GRAYSCALE)

# 若是彩色圖片，可先二值化
# _, img = cv2.threshold(img, 128, 255, cv2.THRESH_BINARY)

# Kernel
kernel = np.ones((5,5), np.uint8)

# 各種形態學
erosion  = cv2.erode(img, kernel, iterations=1)
dilation = cv2.dilate(img, kernel, iterations=1)

opening  = cv2.morphologyEx(img, cv2.MORPH_OPEN, kernel)
closing  = cv2.morphologyEx(img, cv2.MORPH_CLOSE, kernel)

gradient = cv2.morphologyEx(img, cv2.MORPH_GRADIENT, kernel)

tophat   = cv2.morphologyEx(img, cv2.MORPH_TOPHAT, kernel)

blackhat = cv2.morphologyEx(img, cv2.MORPH_BLACKHAT, kernel)

#==============================
# 顯示結果
#==============================

titles = [
    "Original",
    "Erosion",
    "Dilation",
    "Opening",
    "Closing",
    "Gradient",
    "Top Hat",
    "Black Hat"
]

images = [
    img,
    erosion,
    dilation,
    opening,
    closing,
    gradient,
    tophat,
    blackhat
]

plt.figure(figsize=(15,8))

for i in range(len(images)):
    plt.subplot(2,4,i+1)
    plt.imshow(images[i], cmap='gray')
    plt.title(titles[i], fontsize=12)
    plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt


# =========================
# 讀取圖片
# =========================

img = cv2.imread(
    "j.png",
    cv2.IMREAD_GRAYSCALE
)


# 如果不是二值圖，先轉二值
_, binary = cv2.threshold(
    img,
    128,
    255,
    cv2.THRESH_BINARY
)


# Kernel
kernel = np.ones((5,5), np.uint8)



# =========================
# Morphology functions
# =========================

results = {

    "Erosion":
        cv2.erode(
            binary,
            kernel,
            iterations=1
        ),

    "Dilation":
        cv2.dilate(
            binary,
            kernel,
            iterations=1
        ),

    "Opening":
        cv2.morphologyEx(
            binary,
            cv2.MORPH_OPEN,
            kernel
        ),

    "Closing":
        cv2.morphologyEx(
            binary,
            cv2.MORPH_CLOSE,
            kernel
        ),

    "Gradient":
        cv2.morphologyEx(
            binary,
            cv2.MORPH_GRADIENT,
            kernel
        ),

    "Top Hat":
        cv2.morphologyEx(
            binary,
            cv2.MORPH_TOPHAT,
            kernel
        ),

    "Black Hat":
        cv2.morphologyEx(
            binary,
            cv2.MORPH_BLACKHAT,
            kernel
        )
}



# =========================
# 一種操作一張圖
# 上：Original
# 下：Result
# =========================

for name, result in results.items():

    plt.figure(figsize=(6,8))


    # 上面：原圖
    plt.subplot(2,1,1)
    plt.imshow(binary, cmap="gray")
    plt.title(
        f"Original Image",
        fontsize=14
    )
    plt.axis("off")


    # 下面：結果
    plt.subplot(2,1,2)
    plt.imshow(result, cmap="gray")
    plt.title(
        f"{name}",
        fontsize=14
    )
    plt.axis("off")


    plt.tight_layout()
    plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt


# =========================
# 讀取圖片
# =========================

img = cv2.imread(
    "j.png",
    cv2.IMREAD_GRAYSCALE
)


# 二值化
_, binary = cv2.threshold(
    img,
    128,
    255,
    cv2.THRESH_BINARY
)


img_open = cv2.imread(
    "open_org.png",
    cv2.IMREAD_GRAYSCALE
)

img_close = cv2.imread(
    "close_org.png",
    cv2.IMREAD_GRAYSCALE
)


# Kernel
kernel = np.ones((5,5), np.uint8)



# =========================
# Morphology
# =========================



results = {

    "Erosion":
        cv2.erode(
            binary,
            kernel,
            iterations=1
        ),

    "Dilation":
        cv2.dilate(
            binary,
            kernel,
            iterations=1
        ),

    "Opening":
        cv2.morphologyEx(
            img_open,
            cv2.MORPH_OPEN,
            kernel
        ),

    "Closing":
        cv2.morphologyEx(
            img_close,
            cv2.MORPH_CLOSE,
            kernel
        ),

    "Gradient":
        cv2.morphologyEx(
            binary,
            cv2.MORPH_GRADIENT,
            kernel
        ),

    "Top Hat":
        cv2.morphologyEx(
            img_open,
            cv2.MORPH_TOPHAT,
            kernel
        ),

    "Black Hat":
        cv2.morphologyEx(
            img_close,
            cv2.MORPH_BLACKHAT,
            kernel
        )
}



# =========================
# 橫向 Pair 顯示
# Original | Result
# =========================

for name, result in results.items():

    plt.figure(figsize=(10,4))


    # 左：原圖
    plt.subplot(1,2,1)
    if name in ["Opening", "Top Hat"]:
        plt.imshow(img_open, cmap="gray")
    elif name in ["Closing", "Black Hat"]:
        plt.imshow(img_close, cmap="gray")
    else:
        plt.imshow(binary, cmap="gray")
    plt.title("Original")
    plt.axis("off")


    # 右：結果
    plt.subplot(1,2,2)
    plt.imshow(result, cmap="gray")
    plt.title(name)
    plt.axis("off")


    plt.tight_layout()
    plt.show()
```
